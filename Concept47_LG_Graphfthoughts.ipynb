```python
"""
=============================================================
FILE 48 — GRAPH OF THOUGHTS (GoT)
=============================================================

CONCEPTS TAUGHT
----------------
1. Graph of Thoughts
2. Complex Reasoning Networks
3. Non-Linear Reasoning
4. Connected Thought Systems
5. Multi-Hop Reasoning
6. Advanced Deliberation
7. Reasoning Graphs
8. Interconnected Intelligence
9. Dynamic Thought Navigation
10. Advanced Agent Architectures

CORE IDEA
-----------
Instead of linear reasoning,
thoughts are interconnected
like a graph.

Unlike:
- Chain of Thought
- Tree of Thoughts

Graph of Thoughts enables:
- interconnected reasoning
- multi-hop thought exploration
- dynamic traversal

FLOW
-----
Thought A ↔ Thought B
     ↘     ↗
      Thought C
           ↓
Final Synthesis

REAL WORLD USE CASES
---------------------
- strategic reasoning
- advanced planning
- research systems
- complex enterprise analysis
"""

# ============================================================
# STEP 1 — IMPORTS
# ============================================================

import os

from dotenv import load_dotenv

from typing_extensions import TypedDict

from langgraph.graph import StateGraph, START, END

from IPython.display import Image, display

# ============================================================
# STEP 2 — LOAD ENV
# ============================================================

load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")

# ============================================================
# STEP 3 — MODEL
# ============================================================

from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o-mini")

# ============================================================
# STEP 4 — STATE
# ============================================================

class State(TypedDict):

    problem: str

    technical_reasoning: str
    financial_reasoning: str
    operational_reasoning: str

    integrated_strategy: str

# ============================================================
# STEP 5 — REASONING NODES
# ============================================================

def technical_reasoning(state: State):

    response = llm.invoke(
        f"""
        Analyze technically:

        {state['problem']}
        """
    )

    return {
        "technical_reasoning": response.content
    }

def financial_reasoning(state: State):

    response = llm.invoke(
        f"""
        Analyze financially:

        {state['problem']}
        """
    )

    return {
        "financial_reasoning": response.content
    }

def operational_reasoning(state: State):

    response = llm.invoke(
        f"""
        Analyze operationally:

        {state['problem']}
        """
    )

    return {
        "operational_reasoning": response.content
    }

# ============================================================
# STEP 6 — SYNTHESIS NODE
# ============================================================

def integrated_reasoning(state: State):

    response = llm.invoke(
        f"""
        Build one integrated strategy using:

        TECHNICAL:
        {state['technical_reasoning']}

        FINANCIAL:
        {state['financial_reasoning']}

        OPERATIONAL:
        {state['operational_reasoning']}
        """
    )

    return {
        "integrated_strategy": response.content
    }

# ============================================================
# STEP 7 — BUILD GRAPH
# ============================================================

builder = StateGraph(State)

builder.add_node(
    "technical_reasoning",
    technical_reasoning
)

builder.add_node(
    "financial_reasoning",
    financial_reasoning
)

builder.add_node(
    "operational_reasoning",
    operational_reasoning
)

builder.add_node(
    "integrated_reasoning",
    integrated_reasoning
)

# ============================================================
# STEP 8 — GRAPH STYLE REASONING
# ============================================================

builder.add_edge(START, "technical_reasoning")
builder.add_edge(START, "financial_reasoning")
builder.add_edge(START, "operational_reasoning")

builder.add_edge(
    "technical_reasoning",
    "integrated_reasoning"
)

builder.add_edge(
    "financial_reasoning",
    "integrated_reasoning"
)

builder.add_edge(
    "operational_reasoning",
    "integrated_reasoning"
)

builder.add_edge(
    "integrated_reasoning",
    END
)

# ============================================================
# STEP 9 — COMPILE GRAPH
# ============================================================

graph = builder.compile()

display(
    Image(
        graph.get_graph().draw_mermaid_png()
    )
)

# ============================================================
# STEP 10 — RUN WORKFLOW
# ============================================================

result = graph.invoke(
    {
        "problem":
        """
        Build a national AI strategy
        for a developing country.
        """
    }
)

# ============================================================
# STEP 11 — PRINT RESULT
# ============================================================

print("\nGRAPH OF THOUGHTS STRATEGY\n")
print("=" * 60)

print(result["integrated_strategy"])
```
