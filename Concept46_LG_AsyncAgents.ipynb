```python
"""
=============================================================
FILE 47 — ASYNC AGENT SYSTEMS
=============================================================

CONCEPTS TAUGHT
----------------
1. Async Workflows
2. Non-Blocking Execution
3. Async LLM Calls
4. High Throughput Systems
5. Concurrent AI
6. Scalable Agent Systems
7. Async Programming
8. Parallel Async Agents
9. High Performance AI
10. Enterprise Scalability

CORE IDEA
-----------
Async systems allow multiple
tasks to execute concurrently
without blocking.

FLOW
-----
Task A ─┐
Task B ─┼── Concurrent Execution
Task C ─┘

REAL WORLD USE CASES
---------------------
- scalable AI APIs
- high throughput systems
- distributed agents
- enterprise AI infrastructure
"""

# ============================================================
# STEP 1 — IMPORTS
# ============================================================

import os
import asyncio

from dotenv import load_dotenv

from typing_extensions import TypedDict

from langgraph.graph import StateGraph, START, END

from IPython.display import Image, display

# ============================================================
# STEP 2 — LOAD ENV VARIABLES
# ============================================================

load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")

# ============================================================
# STEP 3 — MODEL
# ============================================================

from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o-mini")

# ============================================================
# STEP 4 — STATE
# ============================================================

class State(TypedDict):

    topic: str

    finance_analysis: str
    marketing_analysis: str
    operations_analysis: str

# ============================================================
# STEP 5 — ASYNC NODES
# ============================================================

async def finance_agent(state: State):

    response = await llm.ainvoke(
        f"""
        Analyze from FINANCE perspective:
        {state['topic']}
        """
    )

    return {
        "finance_analysis": response.content
    }

async def marketing_agent(state: State):

    response = await llm.ainvoke(
        f"""
        Analyze from MARKETING perspective:
        {state['topic']}
        """
    )

    return {
        "marketing_analysis": response.content
    }

async def operations_agent(state: State):

    response = await llm.ainvoke(
        f"""
        Analyze from OPERATIONS perspective:
        {state['topic']}
        """
    )

    return {
        "operations_analysis": response.content
    }

# ============================================================
# STEP 6 — BUILD GRAPH
# ============================================================

builder = StateGraph(State)

builder.add_node("finance_agent", finance_agent)
builder.add_node("marketing_agent", marketing_agent)
builder.add_node("operations_agent", operations_agent)

# ============================================================
# STEP 7 — PARALLEL EXECUTION
# ============================================================

builder.add_edge(START, "finance_agent")
builder.add_edge(START, "marketing_agent")
builder.add_edge(START, "operations_agent")

builder.add_edge("finance_agent", END)
builder.add_edge("marketing_agent", END)
builder.add_edge("operations_agent", END)

# ============================================================
# STEP 8 — COMPILE
# ============================================================

graph = builder.compile()

display(
    Image(
        graph.get_graph().draw_mermaid_png()
    )
)

# ============================================================
# STEP 9 — RUN ASYNC WORKFLOW
# ============================================================

async def run_async_workflow():

    result = await graph.ainvoke(
        {
            "topic":
            """
            AI transformation in banking sector
            """
        }
    )

    print("\nASYNC RESULTS\n")
    print("=" * 60)

    print(result)

asyncio.run(run_async_workflow())
```
