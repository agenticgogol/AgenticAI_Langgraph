```python
# ============================================================
# HUMAN FEEDBACK LEARNING AGENT
# VERSION 2 — AGENT IMPROVES USING FEEDBACK
# ============================================================

"""
============================================================
BIG IMPROVEMENT IN THIS VERSION
============================================================

Previous version:
-----------------
Stored feedback in memory

BUT

Did not CLEARLY demonstrate:
- learning
- behavioral improvement
- response adaptation

============================================================
THIS VERSION SHOWS
============================================================

ROUND 1
--------
AI gives generic answer

Human Feedback:
"I prefer step-by-step roadmap with practical projects"

SYSTEM LEARNS:
--------------
- user likes structured learning
- user likes projects
- user dislikes generic answers

ROUND 2
--------
AI now generates:
- structured roadmap
- project-based guidance
- practical examples

This CLEARLY demonstrates:

HUMAN FEEDBACK
      ↓
MEMORY UPDATE
      ↓
FUTURE BEHAVIOR CHANGE

============================================================
IMPORTANT AGENTIC AI CONCEPT
============================================================

This is a SIMPLE form of:

RLHF-like adaptation
(Reinforcement Learning from Human Feedback)

except implemented through:
- memory
- retrieval
- prompting

instead of model weight updates.

============================================================
"""

# ============================================================
# IMPORTS
# ============================================================

from dotenv import load_dotenv
load_dotenv()

import os
import json
import uuid
import logging

from datetime import datetime

from typing import List

from typing_extensions import TypedDict

from pydantic import BaseModel, Field

from langchain_openai import ChatOpenAI

from langgraph.graph import StateGraph
from langgraph.graph import START, END

from IPython.display import Image, display

import chromadb

from sentence_transformers import SentenceTransformer

# ============================================================
# LOGGING
# ============================================================

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s"
)

logger = logging.getLogger(__name__)

# ============================================================
# OPENAI CONFIG
# ============================================================

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")

# ============================================================
# LLM
# ============================================================

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

# ============================================================
# EMBEDDING MODEL
# ============================================================

embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

# ============================================================
# CHROMADB
# ============================================================

CHROMA_PATH = "./human_feedback_memory_db"

os.makedirs(CHROMA_PATH, exist_ok=True)

chroma_client = chromadb.PersistentClient(
    path=CHROMA_PATH
)

collection = chroma_client.get_or_create_collection(
    name="feedback_learning_agent"
)

# ============================================================
# MEMORY FILE
# ============================================================

MEMORY_FILE = "feedback_learning_memory.json"

# ============================================================
# INITIALIZE MEMORY
# ============================================================

def initialize_memory():

    if not os.path.exists(MEMORY_FILE):

        memory = {

            "preferences": {},

            "feedbacks": {},

            "conversation_history": {}

        }

        with open(MEMORY_FILE, "w") as f:

            json.dump(memory, f, indent=4)

initialize_memory()

# ============================================================
# LOAD MEMORY
# ============================================================

def load_memory():

    with open(MEMORY_FILE, "r") as f:

        return json.load(f)

# ============================================================
# SAVE MEMORY
# ============================================================

def save_memory(memory):

    with open(MEMORY_FILE, "w") as f:

        json.dump(memory, f, indent=4)

# ============================================================
# STORE MEMORY
# ============================================================

def store_memory(memory_type, user_id, item):

    memory = load_memory()

    if user_id not in memory[memory_type]:

        memory[memory_type][user_id] = []

    memory[memory_type][user_id].append(item)

    save_memory(memory)

    logger.info(
        f"Stored in {memory_type}: {item}"
    )

# ============================================================
# GET MEMORY
# ============================================================

def get_memory(memory_type, user_id):

    memory = load_memory()

    return memory[memory_type].get(user_id, [])

# ============================================================
# VECTOR STORE
# ============================================================

def store_vector_memory(user_id, text):

    embedding = embedding_model.encode(
        text
    ).tolist()

    collection.add(

        ids=[str(uuid.uuid4())],

        documents=[text],

        embeddings=[embedding],

        metadatas=[{
            "user_id": user_id,
            "timestamp": str(datetime.now())
        }]
    )

# ============================================================
# VECTOR RETRIEVAL
# ============================================================

def retrieve_vector_memory(query, top_k=3):

    query_embedding = embedding_model.encode(
        query
    ).tolist()

    results = collection.query(

        query_embeddings=[query_embedding],

        n_results=top_k
    )

    docs = []

    if results["documents"]:

        for group in results["documents"]:

            docs.extend(group)

    return docs

# ============================================================
# FEEDBACK EXTRACTION SCHEMA
# ============================================================

class LearningSchema(BaseModel):

    user_preferences: List[str] = Field(
        description="User preferences"
    )

    disliked_patterns: List[str] = Field(
        description="Things user dislikes"
    )

    improvement_rules: List[str] = Field(
        description="How assistant should improve"
    )

# ============================================================
# FEEDBACK EXTRACTOR
# ============================================================

feedback_extractor = llm.with_structured_output(
    LearningSchema
)

# ============================================================
# STATE
# ============================================================

class State(TypedDict):

    user_id: str

    user_query: str

    retrieved_preferences: str

    retrieved_feedbacks: str

    vector_context: str

    ai_response: str

    human_feedback: str

# ============================================================
# RETRIEVE MEMORY NODE
# ============================================================

def retrieve_memories(state: State):

    logger.info("="*60)
    logger.info("RETRIEVING MEMORIES")
    logger.info("="*60)

    user_id = state["user_id"]

    preferences = get_memory(
        "preferences",
        user_id
    )

    feedbacks = get_memory(
        "feedbacks",
        user_id
    )

    vector_results = retrieve_vector_memory(
        state["user_query"]
    )

    logger.info(
        f"Preferences Retrieved: {len(preferences)}"
    )

    logger.info(
        f"Feedbacks Retrieved: {len(feedbacks)}"
    )

    return {

        "retrieved_preferences":
        "\n".join(preferences),

        "retrieved_feedbacks":
        "\n".join(feedbacks),

        "vector_context":
        "\n".join(vector_results)

    }

# ============================================================
# RESPONSE GENERATION NODE
# ============================================================

def generate_response(state: State):

    logger.info("="*60)
    logger.info("GENERATING RESPONSE")
    logger.info("="*60)

    prompt = f"""
    You are an adaptive AI assistant.

    ==================================================
    USER PREFERENCES
    ==================================================

    {state['retrieved_preferences']}

    ==================================================
    PREVIOUS FEEDBACKS
    ==================================================

    {state['retrieved_feedbacks']}

    ==================================================
    VECTOR MEMORIES
    ==================================================

    {state['vector_context']}

    ==================================================
    USER QUERY
    ==================================================

    {state['user_query']}

    ==================================================
    VERY IMPORTANT
    ==================================================

    Learn from previous feedback.

    Improve your answer style accordingly.

    Avoid previous mistakes.

    Personalize heavily.
    """

    response = llm.invoke(prompt)

    return {
        "ai_response": response.content
    }

# ============================================================
# FEEDBACK COLLECTION NODE
# ============================================================

def collect_feedback(state: State):

    logger.info("="*60)
    logger.info("COLLECTING HUMAN FEEDBACK")
    logger.info("="*60)

    print("\n")
    print("="*80)
    print("AI RESPONSE")
    print("="*80)

    print(state["ai_response"])

    print("\n")
    print("="*80)
    print("HUMAN FEEDBACK")
    print("="*80)

    feedback = input(
        "Enter feedback for the AI response: "
    )

    return {
        "human_feedback": feedback
    }

# ============================================================
# LEARNING NODE
# ============================================================

def learn_from_feedback(state: State):

    logger.info("="*60)
    logger.info("LEARNING FROM FEEDBACK")
    logger.info("="*60)

    user_id = state["user_id"]

    feedback = state["human_feedback"]

    # Store raw feedback

    store_memory(
        "feedbacks",
        user_id,
        feedback
    )

    # Extract learnings

    extracted = feedback_extractor.invoke(
        f"""
        Analyze this feedback carefully.

        Feedback:
        {feedback}

        Extract:
        - user preferences
        - disliked patterns
        - improvement rules
        """
    )

    # Store preferences

    for item in extracted.user_preferences:

        store_memory(
            "preferences",
            user_id,
            item
        )

    # Store improvement rules in vector memory

    for item in extracted.improvement_rules:

        store_vector_memory(
            user_id,
            item
        )

    logger.info("Agent Learning Updated")

    print("\n")
    print("="*80)
    print("LEARNINGS EXTRACTED")
    print("="*80)

    print("\nUSER PREFERENCES:")

    for item in extracted.user_preferences:

        print(f"- {item}")

    print("\nIMPROVEMENT RULES:")

    for item in extracted.improvement_rules:

        print(f"- {item}")

    return {}

# ============================================================
# BUILD GRAPH
# ============================================================

builder = StateGraph(State)

builder.add_node(
    "retrieve_memories",
    retrieve_memories
)

builder.add_node(
    "generate_response",
    generate_response
)

builder.add_node(
    "collect_feedback",
    collect_feedback
)

builder.add_node(
    "learn_from_feedback",
    learn_from_feedback
)

# ============================================================
# EDGES
# ============================================================

builder.add_edge(
    START,
    "retrieve_memories"
)

builder.add_edge(
    "retrieve_memories",
    "generate_response"
)

builder.add_edge(
    "generate_response",
    "collect_feedback"
)

builder.add_edge(
    "collect_feedback",
    "learn_from_feedback"
)

builder.add_edge(
    "learn_from_feedback",
    END
)

# ============================================================
# COMPILE GRAPH
# ============================================================

graph = builder.compile()

# ============================================================
# SHOW GRAPH
# ============================================================

print("\n")
print("="*80)
print("LANGGRAPH WORKFLOW")
print("="*80)

display(
    Image(
        graph.get_graph().draw_mermaid_png()
    )
)

# ============================================================
# ROUND 1
# ============================================================

print("\n")
print("="*80)
print("ROUND 1 — BEFORE LEARNING")
print("="*80)

input_data = {

    "user_id": "utsab",

    "user_query":
    """
    Help me become an AI Architect.
    """
}

result_1 = graph.invoke(input_data)

# ============================================================
# ROUND 2
# ============================================================

"""
NOW THE AGENT HAS LEARNED FROM FEEDBACK

It should improve responses.
"""

print("\n")
print("="*80)
print("ROUND 2 — AFTER LEARNING")
print("="*80)

input_data_2 = {

    "user_id": "utsab",

    "user_query":
    """
    Create a roadmap for becoming
    a Senior AI Leader.
    """
}

# Retrieve updated memories

memory_state = retrieve_memories(input_data_2)

# Generate improved response

improved_response = generate_response({

    **input_data_2,

    **memory_state
})

print("\n")
print("="*80)
print("IMPROVED RESPONSE AFTER FEEDBACK LEARNING")
print("="*80)

print(improved_response["ai_response"])

# ============================================================
# SHOW FINAL MEMORY
# ============================================================

print("\n")
print("="*80)
print("FINAL MEMORY STORE")
print("="*80)

memory_dump = load_memory()

print(
    json.dumps(
        memory_dump,
        indent=4
    )
)
```
