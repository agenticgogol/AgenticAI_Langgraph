```python
"""
=============================================================
FILE 40 — SWARM INTELLIGENCE PATTERN
=============================================================

CONCEPTS TAUGHT
----------------
1. Swarm Intelligence
2. Decentralized AI Systems
3. Collective Intelligence
4. Emergent Behavior
5. Distributed Agents
6. Parallel Collaboration
7. Consensus Building
8. Multi-Agent Coordination
9. Biological Inspiration
10. Autonomous Collaboration

CORE IDEA
-----------
Many lightweight agents independently
analyze a problem and collectively
produce a stronger solution.

Inspired by:
- ants
- bees
- bird flocks
- decentralized systems

FLOW
-----
Problem
   ↓
Multiple Independent Agents
   ↓
Collective Insights
   ↓
Aggregation / Consensus

REAL WORLD USE CASES
---------------------
- Market analysis
- Risk assessment
- Research synthesis
- Distributed AI systems
- Recommendation engines
"""

# ============================================================
# STEP 1 — IMPORTS
# ============================================================

import os

from dotenv import load_dotenv

from typing_extensions import TypedDict

from langgraph.graph import StateGraph, START, END

from IPython.display import Image, display

# ============================================================
# STEP 2 — LOAD ENV VARIABLES
# ============================================================

load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")

# ============================================================
# STEP 3 — INITIALIZE LLM
# ============================================================

from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o-mini")

# ============================================================
# STEP 4 — DEFINE STATE
# ============================================================

class State(TypedDict):

    problem: str

    agent_1_output: str
    agent_2_output: str
    agent_3_output: str
    agent_4_output: str

    collective_decision: str

# ============================================================
# STEP 5 — SWARM AGENTS
# ============================================================

def finance_swarm_agent(state: State):

    response = llm.invoke(
        f"""
        Analyze the following problem
        from FINANCIAL perspective.

        PROBLEM:
        {state['problem']}
        """
    )

    return {
        "agent_1_output": response.content
    }

def marketing_swarm_agent(state: State):

    response = llm.invoke(
        f"""
        Analyze the following problem
        from MARKETING perspective.

        PROBLEM:
        {state['problem']}
        """
    )

    return {
        "agent_2_output": response.content
    }

def operations_swarm_agent(state: State):

    response = llm.invoke(
        f"""
        Analyze the following problem
        from OPERATIONS perspective.

        PROBLEM:
        {state['problem']}
        """
    )

    return {
        "agent_3_output": response.content
    }

def customer_swarm_agent(state: State):

    response = llm.invoke(
        f"""
        Analyze the following problem
        from CUSTOMER EXPERIENCE perspective.

        PROBLEM:
        {state['problem']}
        """
    )

    return {
        "agent_4_output": response.content
    }

# ============================================================
# STEP 6 — COLLECTIVE AGGREGATOR
# ============================================================

def collective_reasoning_node(state: State):

    response = llm.invoke(
        f"""
        Combine all swarm intelligence outputs
        into one enterprise recommendation.

        FINANCE:
        {state['agent_1_output']}

        MARKETING:
        {state['agent_2_output']}

        OPERATIONS:
        {state['agent_3_output']}

        CUSTOMER EXPERIENCE:
        {state['agent_4_output']}
        """
    )

    return {
        "collective_decision": response.content
    }

# ============================================================
# STEP 7 — BUILD GRAPH
# ============================================================

builder = StateGraph(State)

builder.add_node("finance_swarm_agent", finance_swarm_agent)
builder.add_node("marketing_swarm_agent", marketing_swarm_agent)
builder.add_node("operations_swarm_agent", operations_swarm_agent)
builder.add_node("customer_swarm_agent", customer_swarm_agent)

builder.add_node(
    "collective_reasoning_node",
    collective_reasoning_node
)

# ============================================================
# STEP 8 — PARALLEL EXECUTION
# ============================================================

builder.add_edge(START, "finance_swarm_agent")
builder.add_edge(START, "marketing_swarm_agent")
builder.add_edge(START, "operations_swarm_agent")
builder.add_edge(START, "customer_swarm_agent")

builder.add_edge(
    "finance_swarm_agent",
    "collective_reasoning_node"
)

builder.add_edge(
    "marketing_swarm_agent",
    "collective_reasoning_node"
)

builder.add_edge(
    "operations_swarm_agent",
    "collective_reasoning_node"
)

builder.add_edge(
    "customer_swarm_agent",
    "collective_reasoning_node"
)

builder.add_edge(
    "collective_reasoning_node",
    END
)

# ============================================================
# STEP 9 — COMPILE GRAPH
# ============================================================

graph = builder.compile()

display(
    Image(
        graph.get_graph().draw_mermaid_png()
    )
)

# ============================================================
# STEP 10 — RUN WORKFLOW
# ============================================================

result = graph.invoke(
    {
        "problem":
        """
        Should a retail bank aggressively adopt
        AI-powered customer service systems?
        """
    }
)

# ============================================================
# STEP 11 — PRINT RESULT
# ============================================================

print("\nCOLLECTIVE SWARM DECISION\n")
print("=" * 60)

print(result["collective_decision"])
```
