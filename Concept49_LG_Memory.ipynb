```python
"""
============================================================
ADVANCED MEMORY AUGMENTED AGENT USING LANGGRAPH
============================================================

This example demonstrates production-style memory systems
inside an Agentic AI application.

We implement:

1. Episodic Memory
   - Stores past interactions/events
   - Example:
     "User asked about leadership roles"

2. Semantic Memory
   - Stores facts/preferences
   - Example:
     "User prefers remote jobs"

3. Procedural Memory
   - Stores workflows / habits / routines
   - Example:
     "User likes step-by-step learning"

4. Vector Memory
   - Semantic retrieval using embeddings
   - Retrieves relevant memories dynamically

============================================================
ARCHITECTURE
============================================================

User Query
    ↓
Memory Retrieval Layer
    ↓
Retrieve:
   - Episodic
   - Semantic
   - Procedural
   - Vector Similar Memories
    ↓
LLM Reasoning
    ↓
Generate Personalized Response
    ↓
Memory Extraction Layer
    ↓
Update Memories
    ↓
Persist Memory Store

============================================================
PRODUCTION FEATURES
============================================================

✅ LangGraph Workflow
✅ Persistent JSON Memory Store
✅ Vector Embeddings
✅ Semantic Search
✅ Structured Memory Extraction
✅ Memory Categorization
✅ Dynamic Retrieval
✅ Memory Updating
✅ Typed State
✅ Modular Architecture
"""
```

```python
# ============================================================
# ADVANCED MEMORY AUGMENTED AGENT
# FIXED VERSION (NO READONLY DATABASE ERROR)
# ============================================================

"""

============================================================
WHAT THIS PROJECT DEMONSTRATES
============================================================

✅ LangGraph Memory Systems
✅ Episodic Memory
✅ Semantic Memory
✅ Procedural Memory
✅ Vector Memory
✅ ChromaDB Semantic Retrieval
✅ Logging
✅ Streaming Execution
✅ Mermaid Visualization
✅ Memory Updates
✅ Personalized AI Agent

============================================================
INSTALL
============================================================

pip install langgraph
pip install langchain-openai
pip install chromadb
pip install sentence-transformers
pip install python-dotenv
pip install ipython

============================================================
"""

# ============================================================
# IMPORTS
# ============================================================

from dotenv import load_dotenv
load_dotenv()

import os
import json
import uuid
import tempfile
import logging

from datetime import datetime

from typing import List
from typing_extensions import TypedDict

from pydantic import BaseModel, Field

from langchain_openai import ChatOpenAI

from langgraph.graph import StateGraph
from langgraph.graph import START, END

from IPython.display import Image, display

# Vector DB

import chromadb

from sentence_transformers import SentenceTransformer

# ============================================================
# LOGGING CONFIG
# ============================================================

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s"
)

logger = logging.getLogger(__name__)

# ============================================================
# OPENAI CONFIG
# ============================================================

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")

# ============================================================
# LLM
# ============================================================

logger.info("Initializing OpenAI Model...")

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

logger.info("LLM Initialized")

# ============================================================
# EMBEDDING MODEL
# ============================================================

logger.info("Loading Embedding Model...")

embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

logger.info("Embedding Model Loaded")

# ============================================================
# FIXED CHROMADB SETUP
# ============================================================

"""
IMPORTANT FIX

Instead of:
Persistent local folder

We use:
tempfile.mkdtemp()

This avoids readonly DB issues.
"""

logger.info("Initializing ChromaDB...")

CHROMA_PATH = tempfile.mkdtemp()

logger.info(f"Temporary Chroma Path: {CHROMA_PATH}")

chroma_client = chromadb.PersistentClient(
    path=CHROMA_PATH
)

collection = chroma_client.get_or_create_collection(
    name="agent_memory"
)

logger.info("ChromaDB Ready")

# ============================================================
# MEMORY FILE
# ============================================================

MEMORY_FILE = "advanced_memory.json"

# ============================================================
# INITIALIZE MEMORY STORE
# ============================================================

def initialize_memory_store():

    logger.info("Initializing Memory Store...")

    if not os.path.exists(MEMORY_FILE):

        memory_structure = {

            "episodic_memory": {},

            "semantic_memory": {},

            "procedural_memory": {}

        }

        with open(MEMORY_FILE, "w") as f:

            json.dump(memory_structure, f, indent=4)

        logger.info("Memory Store Created")

    else:

        logger.info("Memory Store Already Exists")

initialize_memory_store()

# ============================================================
# LOAD MEMORY
# ============================================================

def load_memory():

    with open(MEMORY_FILE, "r") as f:

        return json.load(f)

# ============================================================
# SAVE MEMORY
# ============================================================

def save_memory(memory):

    with open(MEMORY_FILE, "w") as f:

        json.dump(memory, f, indent=4)

# ============================================================
# MEMORY EXTRACTION SCHEMA
# ============================================================

class ExtractedMemory(BaseModel):

    episodic: List[str] = Field(
        description="Past events and experiences"
    )

    semantic: List[str] = Field(
        description="Facts and preferences"
    )

    procedural: List[str] = Field(
        description="Habits and workflows"
    )

# ============================================================
# STRUCTURED MEMORY EXTRACTOR
# ============================================================

memory_extractor = llm.with_structured_output(
    ExtractedMemory
)

# ============================================================
# STATE
# ============================================================

class State(TypedDict):

    user_id: str

    user_query: str

    episodic_context: str

    semantic_context: str

    procedural_context: str

    vector_context: str

    final_response: str

# ============================================================
# GET MEMORY
# ============================================================

def get_user_memory(memory_type, user_id):

    memory = load_memory()

    return memory[memory_type].get(user_id, [])

# ============================================================
# STORE MEMORY
# ============================================================

def store_memory(memory_type, user_id, item):

    logger.info(f"Storing {memory_type}: {item}")

    memory = load_memory()

    if user_id not in memory[memory_type]:

        memory[memory_type][user_id] = []

    if item not in memory[memory_type][user_id]:

        memory[memory_type][user_id].append(item)

    save_memory(memory)

# ============================================================
# STORE VECTOR MEMORY
# ============================================================

def store_vector_memory(user_id, text):

    logger.info("Updating Vector Memory...")

    try:

        embedding = embedding_model.encode(
            text
        ).tolist()

        collection.add(

            ids=[str(uuid.uuid4())],

            documents=[text],

            embeddings=[embedding],

            metadatas=[{
                "user_id": user_id,
                "timestamp": str(datetime.now())
            }]
        )

        logger.info("Vector Memory Stored Successfully")

    except Exception as e:

        logger.error(
            f"Vector Memory Storage Failed: {e}"
        )

# ============================================================
# RETRIEVE VECTOR MEMORY
# ============================================================

def retrieve_vector_memory(user_id, query, top_k=3):

    logger.info("Performing Semantic Search...")

    try:

        query_embedding = embedding_model.encode(
            query
        ).tolist()

        results = collection.query(

            query_embeddings=[query_embedding],

            n_results=top_k
        )

        retrieved_docs = []

        if results["documents"]:

            for doc_group in results["documents"]:

                retrieved_docs.extend(doc_group)

        logger.info(
            f"Retrieved {len(retrieved_docs)} Vector Memories"
        )

        return retrieved_docs

    except Exception as e:

        logger.error(
            f"Vector Retrieval Failed: {e}"
        )

        return []

# ============================================================
# MEMORY RETRIEVAL NODE
# ============================================================

def retrieve_memories(state: State):

    logger.info("="*60)
    logger.info("STARTING MEMORY RETRIEVAL")
    logger.info("="*60)

    logger.info(
        f"User Query: {state['user_query']}"
    )

    user_id = state["user_id"]

    # Episodic

    episodic = get_user_memory(
        "episodic_memory",
        user_id
    )

    logger.info(
        f"Episodic Memories Found: {len(episodic)}"
    )

    # Semantic

    semantic = get_user_memory(
        "semantic_memory",
        user_id
    )

    logger.info(
        f"Semantic Memories Found: {len(semantic)}"
    )

    # Procedural

    procedural = get_user_memory(
        "procedural_memory",
        user_id
    )

    logger.info(
        f"Procedural Memories Found: {len(procedural)}"
    )

    # Vector

    vector_results = retrieve_vector_memory(
        user_id,
        state["user_query"]
    )

    logger.info("Memory Retrieval Completed")

    return {

        "episodic_context":
        "\n".join(episodic[-5:]),

        "semantic_context":
        "\n".join(semantic[-5:]),

        "procedural_context":
        "\n".join(procedural[-5:]),

        "vector_context":
        "\n".join(vector_results)

    }

# ============================================================
# REASONING NODE
# ============================================================

def reasoning_agent(state: State):

    logger.info("="*60)
    logger.info("STARTING REASONING AGENT")
    logger.info("="*60)

    prompt = f"""
    You are an advanced AI Career Coach.

    ==================================================
    EPISODIC MEMORY
    ==================================================

    {state['episodic_context']}

    ==================================================
    SEMANTIC MEMORY
    ==================================================

    {state['semantic_context']}

    ==================================================
    PROCEDURAL MEMORY
    ==================================================

    {state['procedural_context']}

    ==================================================
    VECTOR MEMORY
    ==================================================

    {state['vector_context']}

    ==================================================
    USER QUERY
    ==================================================

    {state['user_query']}

    ==================================================
    TASK
    ==================================================

    Give deeply personalized guidance.
    """

    logger.info("Calling OpenAI...")

    response = llm.invoke(prompt)

    logger.info("LLM Response Generated")

    return {
        "final_response": response.content
    }

# ============================================================
# MEMORY UPDATE NODE
# ============================================================

def update_memories(state: State):

    logger.info("="*60)
    logger.info("UPDATING MEMORIES")
    logger.info("="*60)

    user_id = state["user_id"]

    logger.info("Extracting Structured Memories...")

    extracted = memory_extractor.invoke(
        f"""
        Extract memory from:

        {state['user_query']}

        Categorize into:
        - episodic
        - semantic
        - procedural
        """
    )

    logger.info(
        f"Episodic Extracted: {len(extracted.episodic)}"
    )

    logger.info(
        f"Semantic Extracted: {len(extracted.semantic)}"
    )

    logger.info(
        f"Procedural Extracted: {len(extracted.procedural)}"
    )

    # Episodic

    for item in extracted.episodic:

        store_memory(
            "episodic_memory",
            user_id,
            item
        )

    # Semantic

    for item in extracted.semantic:

        store_memory(
            "semantic_memory",
            user_id,
            item
        )

    # Procedural

    for item in extracted.procedural:

        store_memory(
            "procedural_memory",
            user_id,
            item
        )

    # Vector

    store_vector_memory(
        user_id,
        state["user_query"]
    )

    logger.info("Memory Update Completed")

    return {}

# ============================================================
# BUILD GRAPH
# ============================================================

logger.info("Building LangGraph Workflow...")

builder = StateGraph(State)

builder.add_node(
    "retrieve_memories",
    retrieve_memories
)

builder.add_node(
    "reasoning_agent",
    reasoning_agent
)

builder.add_node(
    "update_memories",
    update_memories
)

# ============================================================
# EDGES
# ============================================================

builder.add_edge(
    START,
    "retrieve_memories"
)

builder.add_edge(
    "retrieve_memories",
    "reasoning_agent"
)

builder.add_edge(
    "reasoning_agent",
    "update_memories"
)

builder.add_edge(
    "update_memories",
    END
)

# ============================================================
# COMPILE GRAPH
# ============================================================

graph = builder.compile()

logger.info("LangGraph Compiled Successfully")

# ============================================================
# SHOW MERMAID GRAPH
# ============================================================

print("\n")
print("="*80)
print("LANGGRAPH WORKFLOW")
print("="*80)

display(
    Image(
        graph.get_graph().draw_mermaid_png()
    )
)

# ============================================================
# INITIAL MEMORY
# ============================================================

logger.info("Initializing User Memories...")

store_memory(
    "semantic_memory",
    "utsab",
    "User has 15 years experience in analytics"
)

store_memory(
    "semantic_memory",
    "utsab",
    "User wants leadership role in AI"
)

store_memory(
    "procedural_memory",
    "utsab",
    "User prefers structured learning"
)

store_memory(
    "episodic_memory",
    "utsab",
    "User previously discussed multi-agent systems"
)

store_vector_memory(
    "utsab",
    "User teaches Agentic AI and LangGraph"
)

# ============================================================
# INPUT
# ============================================================

input_data = {

    "user_id": "utsab",

    "user_query":
    """
    I want to transition into a senior AI leadership role.
    What should I learn next?
    """
}

# ============================================================
# STREAM EXECUTION
# ============================================================

print("\n")
print("="*80)
print("STREAMING GRAPH EXECUTION")
print("="*80)

for event in graph.stream(input_data):

    print("\n")
    print("="*60)
    print("GRAPH EVENT")
    print("="*60)

    print(event)

# ============================================================
# FINAL EXECUTION
# ============================================================

result = graph.invoke(input_data)

# ============================================================
# FINAL RESPONSE
# ============================================================

print("\n")
print("="*80)
print("FINAL PERSONALIZED RESPONSE")
print("="*80)

print("\n")

print(result["final_response"])

# ============================================================
# SHOW MEMORY STORE
# ============================================================

print("\n")
print("="*80)
print("FINAL MEMORY STORE")
print("="*80)

memory_dump = load_memory()

print(
    json.dumps(
        memory_dump,
        indent=4
    )
)
```

```python
# ============================================================
# ADVANCED MEMORY AUGMENTED AGENT
# FIXED VERSION (NO READONLY DATABASE ERROR)
# ============================================================

"""
============================================================
WHAT THIS PROJECT DEMONSTRATES
============================================================

✅ LangGraph Memory Systems
✅ Episodic Memory
✅ Semantic Memory
✅ Procedural Memory
✅ Vector Memory
✅ ChromaDB Semantic Retrieval
✅ Logging
✅ Streaming Execution
✅ Mermaid Visualization
✅ Memory Updates
✅ Personalized AI Agent

============================================================
INSTALL
============================================================

pip install langgraph
pip install langchain-openai
pip install chromadb
pip install sentence-transformers
pip install python-dotenv
pip install ipython

============================================================
"""

# ============================================================
# IMPORTS
# ============================================================

from dotenv import load_dotenv
load_dotenv()

import os
import json
import uuid
import tempfile
import logging

from datetime import datetime

from typing import List
from typing_extensions import TypedDict

from pydantic import BaseModel, Field

from langchain_openai import ChatOpenAI

from langgraph.graph import StateGraph
from langgraph.graph import START, END

from IPython.display import Image, display

# Vector DB

import chromadb

from sentence_transformers import SentenceTransformer

# ============================================================
# LOGGING CONFIG
# ============================================================

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s"
)

logger = logging.getLogger(__name__)

# ============================================================
# OPENAI CONFIG
# ============================================================

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")

# ============================================================
# LLM
# ============================================================

logger.info("Initializing OpenAI Model...")

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

logger.info("LLM Initialized")

# ============================================================
# EMBEDDING MODEL
# ============================================================

logger.info("Loading Embedding Model...")

embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

logger.info("Embedding Model Loaded")

# ============================================================
# FIXED CHROMADB SETUP
# ============================================================

# ============================================================
# PERSISTENT CHROMADB SETUP
# ============================================================

logger.info("Initializing Persistent ChromaDB...")

# Persistent folder path

CHROMA_PATH = "./persistent_chroma_db"

# Create folder if it does not exist

os.makedirs(CHROMA_PATH, exist_ok=True)

# Show absolute path

logger.info(
    f"Persistent Chroma Path: {os.path.abspath(CHROMA_PATH)}"
)

print("\n")
print("="*80)
print("PERSISTENT CHROMADB PATH")
print("="*80)

print(os.path.abspath(CHROMA_PATH))

# Initialize persistent client

chroma_client = chromadb.PersistentClient(
    path=CHROMA_PATH
)

# Create collection

collection = chroma_client.get_or_create_collection(
    name="agent_memory"
)

logger.info("Persistent ChromaDB Ready")

# ============================================================
# MEMORY FILE
# ============================================================

MEMORY_FILE = "advanced_memory.json"

# ============================================================
# INITIALIZE MEMORY STORE
# ============================================================

def initialize_memory_store():

    logger.info("Initializing Memory Store...")

    if not os.path.exists(MEMORY_FILE):

        memory_structure = {

            "episodic_memory": {},

            "semantic_memory": {},

            "procedural_memory": {}

        }

        with open(MEMORY_FILE, "w") as f:

            json.dump(memory_structure, f, indent=4)

        logger.info("Memory Store Created")

    else:

        logger.info("Memory Store Already Exists")

initialize_memory_store()

# ============================================================
# LOAD MEMORY
# ============================================================

def load_memory():

    with open(MEMORY_FILE, "r") as f:

        return json.load(f)

# ============================================================
# SAVE MEMORY
# ============================================================

def save_memory(memory):

    with open(MEMORY_FILE, "w") as f:

        json.dump(memory, f, indent=4)

# ============================================================
# MEMORY EXTRACTION SCHEMA
# ============================================================

class ExtractedMemory(BaseModel):

    episodic: List[str] = Field(
        description="Past events and experiences"
    )

    semantic: List[str] = Field(
        description="Facts and preferences"
    )

    procedural: List[str] = Field(
        description="Habits and workflows"
    )

# ============================================================
# STRUCTURED MEMORY EXTRACTOR
# ============================================================

memory_extractor = llm.with_structured_output(
    ExtractedMemory
)

# ============================================================
# STATE
# ============================================================

class State(TypedDict):

    user_id: str

    user_query: str

    episodic_context: str

    semantic_context: str

    procedural_context: str

    vector_context: str

    final_response: str

# ============================================================
# GET MEMORY
# ============================================================

def get_user_memory(memory_type, user_id):

    memory = load_memory()

    return memory[memory_type].get(user_id, [])

# ============================================================
# STORE MEMORY
# ============================================================

def store_memory(memory_type, user_id, item):

    logger.info(f"Storing {memory_type}: {item}")

    memory = load_memory()

    if user_id not in memory[memory_type]:

        memory[memory_type][user_id] = []

    if item not in memory[memory_type][user_id]:

        memory[memory_type][user_id].append(item)

    save_memory(memory)

# ============================================================
# STORE VECTOR MEMORY
# ============================================================

def store_vector_memory(user_id, text):

    logger.info("Updating Vector Memory...")

    try:

        embedding = embedding_model.encode(
            text
        ).tolist()

        collection.add(

            ids=[str(uuid.uuid4())],

            documents=[text],

            embeddings=[embedding],

            metadatas=[{
                "user_id": user_id,
                "timestamp": str(datetime.now())
            }]
        )

        logger.info("Vector Memory Stored Successfully")

    except Exception as e:

        logger.error(
            f"Vector Memory Storage Failed: {e}"
        )

# ============================================================
# RETRIEVE VECTOR MEMORY
# ============================================================

def retrieve_vector_memory(user_id, query, top_k=3):

    logger.info("Performing Semantic Search...")

    try:

        query_embedding = embedding_model.encode(
            query
        ).tolist()

        results = collection.query(

            query_embeddings=[query_embedding],

            n_results=top_k
        )

        retrieved_docs = []

        if results["documents"]:

            for doc_group in results["documents"]:

                retrieved_docs.extend(doc_group)

        logger.info(
            f"Retrieved {len(retrieved_docs)} Vector Memories"
        )

        return retrieved_docs

    except Exception as e:

        logger.error(
            f"Vector Retrieval Failed: {e}"
        )

        return []

# ============================================================
# MEMORY RETRIEVAL NODE
# ============================================================

def retrieve_memories(state: State):

    logger.info("="*60)
    logger.info("STARTING MEMORY RETRIEVAL")
    logger.info("="*60)

    logger.info(
        f"User Query: {state['user_query']}"
    )

    user_id = state["user_id"]

    # Episodic

    episodic = get_user_memory(
        "episodic_memory",
        user_id
    )

    logger.info(
        f"Episodic Memories Found: {len(episodic)}"
    )

    # Semantic

    semantic = get_user_memory(
        "semantic_memory",
        user_id
    )

    logger.info(
        f"Semantic Memories Found: {len(semantic)}"
    )

    # Procedural

    procedural = get_user_memory(
        "procedural_memory",
        user_id
    )

    logger.info(
        f"Procedural Memories Found: {len(procedural)}"
    )

    # Vector

    vector_results = retrieve_vector_memory(
        user_id,
        state["user_query"]
    )

    logger.info("Memory Retrieval Completed")

    return {

        "episodic_context":
        "\n".join(episodic[-5:]),

        "semantic_context":
        "\n".join(semantic[-5:]),

        "procedural_context":
        "\n".join(procedural[-5:]),

        "vector_context":
        "\n".join(vector_results)

    }

# ============================================================
# REASONING NODE
# ============================================================

def reasoning_agent(state: State):

    logger.info("="*60)
    logger.info("STARTING REASONING AGENT")
    logger.info("="*60)

    prompt = f"""
    You are an advanced AI Career Coach.

    ==================================================
    EPISODIC MEMORY
    ==================================================

    {state['episodic_context']}

    ==================================================
    SEMANTIC MEMORY
    ==================================================

    {state['semantic_context']}

    ==================================================
    PROCEDURAL MEMORY
    ==================================================

    {state['procedural_context']}

    ==================================================
    VECTOR MEMORY
    ==================================================

    {state['vector_context']}

    ==================================================
    USER QUERY
    ==================================================

    {state['user_query']}

    ==================================================
    TASK
    ==================================================

    Give deeply personalized guidance.
    """

    logger.info("Calling OpenAI...")

    response = llm.invoke(prompt)

    logger.info("LLM Response Generated")

    return {
        "final_response": response.content
    }

# ============================================================
# MEMORY UPDATE NODE
# ============================================================

def update_memories(state: State):

    logger.info("="*60)
    logger.info("UPDATING MEMORIES")
    logger.info("="*60)

    user_id = state["user_id"]

    logger.info("Extracting Structured Memories...")

    extracted = memory_extractor.invoke(
        f"""
        Extract memory from:

        {state['user_query']}

        Categorize into:
        - episodic
        - semantic
        - procedural
        """
    )

    logger.info(
        f"Episodic Extracted: {len(extracted.episodic)}"
    )

    logger.info(
        f"Semantic Extracted: {len(extracted.semantic)}"
    )

    logger.info(
        f"Procedural Extracted: {len(extracted.procedural)}"
    )

    # Episodic

    for item in extracted.episodic:

        store_memory(
            "episodic_memory",
            user_id,
            item
        )

    # Semantic

    for item in extracted.semantic:

        store_memory(
            "semantic_memory",
            user_id,
            item
        )

    # Procedural

    for item in extracted.procedural:

        store_memory(
            "procedural_memory",
            user_id,
            item
        )

    # Vector

    store_vector_memory(
        user_id,
        state["user_query"]
    )

    logger.info("Memory Update Completed")

    return {}

# ============================================================
# BUILD GRAPH
# ============================================================

logger.info("Building LangGraph Workflow...")

builder = StateGraph(State)

builder.add_node(
    "retrieve_memories",
    retrieve_memories
)

builder.add_node(
    "reasoning_agent",
    reasoning_agent
)

builder.add_node(
    "update_memories",
    update_memories
)

# ============================================================
# EDGES
# ============================================================

builder.add_edge(
    START,
    "retrieve_memories"
)

builder.add_edge(
    "retrieve_memories",
    "reasoning_agent"
)

builder.add_edge(
    "reasoning_agent",
    "update_memories"
)

builder.add_edge(
    "update_memories",
    END
)

# ============================================================
# COMPILE GRAPH
# ============================================================

graph = builder.compile()

logger.info("LangGraph Compiled Successfully")

# ============================================================
# SHOW MERMAID GRAPH
# ============================================================

print("\n")
print("="*80)
print("LANGGRAPH WORKFLOW")
print("="*80)

display(
    Image(
        graph.get_graph().draw_mermaid_png()
    )
)

# ============================================================
# INITIAL MEMORY
# ============================================================

logger.info("Initializing User Memories...")

store_memory(
    "semantic_memory",
    "utsab",
    "User has 15 years experience in analytics"
)

store_memory(
    "semantic_memory",
    "utsab",
    "User wants leadership role in AI"
)

store_memory(
    "procedural_memory",
    "utsab",
    "User prefers structured learning"
)

store_memory(
    "episodic_memory",
    "utsab",
    "User previously discussed multi-agent systems"
)

store_vector_memory(
    "utsab",
    "User teaches Agentic AI and LangGraph"
)

# ============================================================
# INPUT
# ============================================================

input_data = {

    "user_id": "utsab",

    "user_query":
    """
    I want to transition into a senior AI leadership role.
    What should I learn next?
    """
}

# ============================================================
# STREAM EXECUTION
# ============================================================

print("\n")
print("="*80)
print("STREAMING GRAPH EXECUTION")
print("="*80)

for event in graph.stream(input_data):

    print("\n")
    print("="*60)
    print("GRAPH EVENT")
    print("="*60)

    print(event)

# ============================================================
# FINAL EXECUTION
# ============================================================

result = graph.invoke(input_data)

# ============================================================
# FINAL RESPONSE
# ============================================================

print("\n")
print("="*80)
print("FINAL PERSONALIZED RESPONSE")
print("="*80)

print("\n")

print(result["final_response"])

# ============================================================
# SHOW MEMORY STORE
# ============================================================

print("\n")
print("="*80)
print("FINAL MEMORY STORE")
print("="*80)

memory_dump = load_memory()

print(
    json.dumps(
        memory_dump,
        indent=4
    )
)
```
